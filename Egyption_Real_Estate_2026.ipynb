## **Cleaning**

```python
# **drop meaningless columns**
sell.drop(columns=['price_period'],inplace=True,errors='ignor')# one value:sell
rent.drop(columns=['price_period'], inplace=True, errors='ignore')# one value:monthly

sell.drop(columns=['offering_type'],inplace=True,errors='ignore')# one value:Residential for Sale
rent.drop(columns=['offering_type'],inplace=True,errors='ignore')# one value:Residential for rent

sell.drop(columns=['category'],inplace=True,errors='ignore')# one value:buy
rent.drop(columns=['category'],inplace=True,errors='ignore')# one value:rent

sell.drop(columns=['rera'],inplace=True,errors='ignore')#NULL
rent.drop(columns=['rera'],inplace=True,errors='ignore')#NULL
```

## **doubted commands**

```python
columns_to_keep = [###########
    
    'property_type',
   
    'completion_status',
    
    'price_egp',
    
    'city',
    'town',
    'district',
    'lat',
    'lon',
    'bedrooms',
    'bathrooms',
    'area_value',
    'furnished',
    'listing_level',
    'is_premium',
    'is_featured',
    'listed_date',
    'payment_method',
    'agent_id',
    'broker_id',
]
sell=sell[columns_to_keep]
rent=rent[columns_to_keep]
```

```python
(sell['area_value'] == 1).sum()
```

```python
(rent['area_value'] == 1).sum()
```

```python
for dataframe in [rent, sell]:
  
  dataframe['listed_date'] = pd.to_datetime(dataframe['listed_date'])

  # Ensure agent_id and broker_id are treated as categorical/string IDs
  dataframe['agent_id'] = (
      dataframe['agent_id'].fillna(-1).astype(int).astype(str)
  )
  dataframe['broker_id'] = (
      dataframe['broker_id'].fillna(-1).astype(int).astype(str)
  )
```

```python
print(f"Rent dupliccated: {rent.duplicated().sum()}")
print(f"Sell dupliccated: {sell.duplicated().sum()}")
```

```python
rent.drop_duplicates(inplace=True)
sell.drop_duplicates(inplace=True)
```

```python
#Filling the completion_status in the rent ste
rent['completion_status'] = rent['completion_status'].fillna('completed')
```

```python
# Filter out unrealistic placeholder areas (< 20 sqm)
sell = sell[sell['area_value'] >= 20].copy()
rent = rent[rent['area_value'] >= 20].copy()
```

```python
#Making bedrooms and bathrooms  Ordered Categorical

bed_order = ['studio', '1', '2', '3', '4', '5', '6', '7', '7+']
bath_order = ['none', '1', '2', '3', '4', '5', '6', '7', '7+']

for df in [sell, rent]:
    df['bedrooms'] = pd.Categorical(df['bedrooms'], categories=bed_order, ordered=True)
    df['bathrooms'] = pd.Categorical(df['bathrooms'], categories=bath_order, ordered=True)
```

```python
sell = sell.dropna(subset=['price_egp']).copy()
```

```python
sell['price_per_sqm'] = sell['price_egp'] / sell['area_value']
sell['price_per_sqm']
```

```python
# In sell: fill NaN with 'NO'
sell['furnished'] = sell['furnished'].fillna('NO')

# In rent: fill NaN with mode or 'Unknown'
rent['furnished'] = rent['furnished'].fillna('YES')
```

```python
sell['district'] = sell['district'].fillna(sell['town'])
rent['district'] = rent['district'].fillna(rent['town'])
```

```python
rent['price_per_sqm'] = rent['price_egp'] / rent['area_value']
```
