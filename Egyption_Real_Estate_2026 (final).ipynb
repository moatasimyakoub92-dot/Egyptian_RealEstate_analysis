## **bedrooms & bathrooms breaking down**

```python
re['bedrooms'].value_counts(dropna=False)
```

```python
re['bathrooms'].value_counts(dropna=False)
```

```python
#Making bedrooms and bathrooms  Ordered Categorical

bed_order = ['studio', '1', '2', '3', '4', '5', '6', '7', '7+']
bath_order = ['none', '1', '2', '3', '4', '5', '6', '7', '7+']

for df in [sell, rent]:
    df['bedrooms'] = pd.Categorical(df['bedrooms'], categories=bed_order, ordered=True)
    df['bathrooms'] = pd.Categorical(df['bathrooms'], categories=bath_order, ordered=True)
```

```python
#Adjusting data types
for dataframe in [rent, sell]:
  
  dataframe['listed_date'] = pd.to_datetime(dataframe['listed_date'])

  # Ensure agent_id and broker_id are treated as categorical/string IDs
  dataframe['agent_id'] = (
      dataframe['agent_id'].fillna(-1).astype(int).astype(str)
  )
  dataframe['broker_id'] = (
      dataframe['broker_id'].fillna(-1).astype(int).astype(str)
  )
```

```python
# temporary dataframe that contains all properties with area <=20
small_sell = sell[sell['area_value'] <= 20].copy()

print("Number of suspicious sale records:", len(small_sell))

print("\nProperty types:")
print(small_sell['property_type'].value_counts())

print("\nArea distribution:")
print(small_sell['area_value'].value_counts().sort_index())
```

```python
pd.crosstab(
    small_sell['property_type'],
    small_sell['area_value']
)
```

```python
#summation of titles in the new dataframe that contains the real area
small_sell['title'].str.extract(
    r'(\d+(?:\.\d+)?)\s*m(?:²|2)?\b',
    expand=False
).notna().sum()
```

```python
#show rows where titles contains the real area
small_sell[
    small_sell['title'].str.extract(
        r'(\d+(?:\.\d+)?)\s*m(?:²|2)?\b',
        expand=False
    ).notna()
][
    ['property_type', 'area_value', 'title']
].head(50)
```

```python
# create new column with the extracted area
small_sell['area_from_title'] = (
    small_sell['title']
    .str.extract(r'(\d+(?:\.\d+)?)\s*m(?:²|2)?\b', expand=False)
    .astype(float))
```

```python
small_sell['area_from_title'].shape # 1457 row (number of area <=20)
```

```python
#the extract is almost true as range of values is normal
print(small_sell.shape)

print(small_sell['area_from_title'].notna().sum())

print(small_sell['area_from_title'].dtype)

print(small_sell['area_from_title'].describe())
```

```python
# check if the most big area values match the property type
small_sell[
    small_sell['area_from_title'] > 500
][
    ['property_type', 'area_value', 'area_from_title', 'title']
]
```

```python
# number of biggest area values
(small_sell['area_from_title'] > 500).sum()
```

 merge the extracted column in the sell dataframe

```python
small_sell.loc[
    small_sell['area_from_title'].notna(),
    'area_value'
] = small_sell.loc[
    small_sell['area_from_title'].notna(),
    'area_from_title'
]
```

```python
small_sell.loc[
    small_sell['area_from_title'].isna(),
    'area_value'
] = np.nan
```

```python
sell['area_status'] = 'Original'
```

```python
# merg 'area_from_title' with 'area_value' in 'area_status'
recovered = small_sell['area_from_title'].notna()

sell.loc[
    small_sell.index[recovered],
    'area_value'
] = small_sell.loc[
    recovered,
    'area_from_title'
]
```

```python
# label area values that coudn't be recovered as 'unknown'
unknown = small_sell['area_from_title'].isna()

sell.loc[
    small_sell.index[unknown],
    'area_value'
] = np.nan

sell.loc[
    small_sell.index[unknown],
    'area_status'
] = 'Unknown'
```

```python
#label 'area_value' that could be recoverd as recovered in 'area_status'
sell.loc[
    small_sell.index[recovered],
    'area_status'
] = 'Recovered'
```
