```python
columns_to_keep = [
    'category',
    'property_type',
    'offering_type',
    'completion_status',
    'price_egp',
    'price_period',
    'city',
    'town',
    'district',
    'lat',
    'lon',
    'bedrooms',
    'bathrooms',
    'area_value',
    'furnished',
    'listing_level',
    'is_premium',
    'is_featured',
    'listed_date',
    'payment_method',
    'agent_id',
    'broker_id',
]
df = df[columns_to_keep]
df
```

```python
df_buy=df[df['category']=='buy']
df_buy
```

```python
df_rent=df[df['category']=='rent']
df_rent
```

```python
df_rent.drop(columns=['price_period'],inplace=True,errors='ignor')
df_buy.drop(columns=['price_period'], inplace=True, errors='ignore')
```

```python
df_rent.head(10)
```

```python
df_rent.drop(columns=['offering_type'],inplace=True,errors='ignore')
```

```python
df_rent.head()
```

```python
df_rent.info()
```

```python
for dataframe in [df_rent, df_buy]:
  # Convert bedrooms and bathrooms to float / numeric
  dataframe['bedrooms'] = pd.to_numeric(dataframe['bedrooms'], errors='coerce')
  dataframe['bathrooms'] = pd.to_numeric(
      dataframe['bathrooms'], errors='coerce'
  )

 
  dataframe['listed_date'] = pd.to_datetime(dataframe['listed_date'])

  # Ensure agent_id and broker_id are treated as categorical/string IDs
  dataframe['agent_id'] = (
      dataframe['agent_id'].fillna(-1).astype(int).astype(str)
  )
  dataframe['broker_id'] = (
      dataframe['broker_id'].fillna(-1).astype(int).astype(str)
  )
```

```python

for dataframe in [df_rent, df_buy]:
  dataframe['furnished'] = dataframe['furnished'].fillna('Unknown')
  dataframe['district'] = dataframe['district'].fillna('Unknown')
 

df_buy['completion_status'] = df_buy['completion_status'].fillna('Unknown')
df_rent['payment_method'] = df_rent['payment_method'].fillna('Unspecified')
```

```python
print(f'Duplicates in Rent dataset: {df_rent.duplicated().sum()}')
print(f'Duplicates in Buy dataset: {df_buy.duplicated().sum()}')
```

```python
df_rent.drop_duplicates(inplace=True)
df_buy.drop_duplicates(inplace=True)
```
